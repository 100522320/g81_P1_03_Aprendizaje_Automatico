## 1 Predicciones para competición

Lo primero que haremos sera importar el mejor modelo, el cual sacamos del notebook de pruebas.

```python
import pandas as pd
import numpy as np
import joblib
from google.colab import files

# DEFINIMOS LA FUNCIÓN arreglar_pdays PARA QUE JOBLIB LA ENCUENTRE AL CARGAR
def arreglar_pdays(X):
    X_nuevo = X.copy()
    if 'pdays' in X_nuevo.columns:
        X_nuevo['pdays_contacted'] = (X_nuevo['pdays'] != -1).astype(int)
        X_nuevo['pdays'] = X_nuevo['pdays'].replace(-1, np.nan)
    return X_nuevo

# Subimos el archivo del modelo
print("Por favor, sube tu archivo 'modelo_final.joblib':")
archivo_modelo = files.upload()

# Obtenemos el nombre exacto del archivo subido
nombre_modelo = list(archivo_modelo.keys())[0]

# Cargamos el modelo usando joblib
modelo_cargado = joblib.load(nombre_modelo)

print(f"¡Modelo '{nombre_modelo}' importado y listo para usar!")
```

 Ahora importaremos los datos de competición y usaremos nuestro modelo para hacer predicciones sobre él y guardarlas en un archivo llamado predicciones.csv:

```python
# Subimos el archivo de datos de la competición
print("Por favor, sube tu archivo 'bank_competition.pkl':")
archivo_datos = files.upload()
nombre_archivo_datos = list(archivo_datos.keys())[0]

# Cargamos los datos en un DataFrame de pandas
df_competicion = pd.read_pickle(nombre_archivo_datos)
print(f"Datos cargados correctamente. Dimensiones: {df_competicion.shape}")

# Realizamos las predicciones con el modelo cargado previamente
predicciones = modelo_cargado.predict(df_competicion)
print("¡Predicciones generadas con éxito!")

# Guardamos las predicciones en un archivo CSV
# Creamos un DataFrame solo con las predicciones
df_predicciones = pd.DataFrame({'deposit_prediction': predicciones})

# Lo guardamos sin el índice para que quede limpio
df_predicciones.to_csv('predicciones.csv', index=False)
print("Archivo 'predicciones.csv' creado correctamente.")
```
